```python
#rutas de trabajo
from pathlib import Path
import pandas as pd
import numpy as np

BASE_DIR = Path("../")
DATOS = BASE_DIR / "datos"

DATOS.mkdir(parents=True, exist_ok=True)
print("Carpeta de datos:", DATOS)
```

```python
# Definir la ruta del archivo CSV
ruta_entrada = DATOS / "dataset_consolidado.csv"

# Cargar el dataset
df = pd.read_csv(ruta_entrada)

# Mostrar los primeros registros
df.head()
```

```python
#revisar dimensiones del dataset
print("Filas y columnas:", df.shape)
print("Cantidad de registros:", df.shape[0])
print("Cantidad de columnas:", df.shape[1])
```

```python
#revisar tipos de datos de cada columna
print("Tipos de datos:")
print(df.dtypes)
print("\nInformación del dataset:")
df.info()
```

```python
#revisar valores nulos
reporte_nulos = pd.DataFrame({
    "cantidad_nulos": df.isnull().sum(),
    "porcentaje_nulos": (df.isnull().mean() * 100).round(2)
})

reporte_nulos
```

```python
#revisar valores duplicados
duplicados = df.duplicated().sum()

print("Registros duplicados:", duplicados)
df[df.duplicated(keep=False)]
```

```python
#normalizar nombres de columnas
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

df.columns
```

```python
#normalizar campos de texto
columnas_texto = df.select_dtypes(include="object").columns

for columna in columnas_texto:
    df[columna] = df[columna].astype("string").str.strip()
print(columnas_texto)
df.head()
```

```python
# Convertir columnas de fecha a tipo datetime

df["fecha_ingreso"] = pd.to_datetime(
    df["fecha_ingreso"],
    errors="coerce"
)

df["fecha_salida"] = pd.to_datetime(
    df["fecha_salida"],
    errors="coerce"
)

df["fecha_evaluacion"] = pd.to_datetime(
    df["fecha_evaluacion"],
    errors="coerce"
)

df["fecha_encuesta"] = pd.to_datetime(
    df["fecha_encuesta"],
    errors="coerce"
)

# Verificar los tipos de datos
print(
    df[
        [
            "fecha_ingreso",
            "fecha_salida",
            "fecha_evaluacion",
            "fecha_encuesta"
        ]
    ].dtypes
)
```

```python
#Corregir valores inválidos

# 1. Validar edad
# Las edades menores de 18 o mayores de 70 se consideran inválidas
df.loc[
    (df["edad"] < 18) | (df["edad"] > 70),
    "edad"
] = np.nan


# 2. Validar horas de capacitación
# No pueden existir cantidades negativas
df.loc[
    df["horas_capacitacion"] < 0,
    "horas_capacitacion"
] = np.nan


# 3. Validar ausencias
# Los días de ausencia no pueden ser negativos
df.loc[
    df["ausencias_dias"] < 0,
    "ausencias_dias"
] = np.nan


# 4. Validar puntaje de desempeño
# Se asume que el puntaje debe estar en un rango de 0 a 100
df.loc[
    (df["puntaje_desempeno"] < 0) |
    (df["puntaje_desempeno"] > 100),
    "puntaje_desempeno"
] = np.nan


# 5. Validar respuestas de encuesta
# La escala utilizada es de 1 a 5
columnas_encuesta = [
    "satisfaccion_laboral",
    "relacion_con_jefatura",
    "reconocimiento",
    "balance_vida_trabajo",
    "intencion_de_permanecer"
]

columnas_encuesta = [
    "respuestas_satisfaccion_laboral",
    "respuestas_relacion_con_jefatura",
    "respuestas_reconocimiento",
    "respuestas_balance_vida_trabajo",
    "respuestas_intencion_de_permanecer"
]

# Convertir valores fuera del rango 1-5 en valores nulos
for columna in columnas_encuesta:
    df.loc[
        (df[columna] < 1) | (df[columna] > 5),
        columna
    ] = np.nan


# 6. Validar fechas
# Una fecha de salida no debería ser anterior a la fecha de ingreso
df.loc[
    df["fecha_salida"] < df["fecha_ingreso"],
    "fecha_salida"
] = pd.NaT


# 7. Revisar cuántos valores nulos quedaron luego de corregir inválidos
print("Valores nulos después de corregir datos inválidos:")
print(df.isnull().sum())
```

```python
#Tratamiento de valores nulos
# Revisar nuevamente la cantidad de valores nulos por columna
print("Valores nulos antes del tratamiento:")
print(df.isnull().sum())


# La columna fecha_salida puede contener valores nulos de forma válida.
# Un valor nulo indica que el colaborador aún no ha salido de la organización.
# Por esta razón, estos valores no serán reemplazados.

print("\nValores nulos en fecha_salida:")
print(df["fecha_salida"].isnull().sum())


# Verificar si existen valores nulos en las demás columnas
columnas_sin_fecha_salida = df.columns.drop("fecha_salida")

print("\nValores nulos en las demás columnas:")
print(df[columnas_sin_fecha_salida].isnull().sum())
```

```python
#Eliminar registros duplicados

# Contar registros duplicados antes de eliminarlos
duplicados_antes = df.duplicated().sum()

print("Registros duplicados antes de limpiar:", duplicados_antes)


# Eliminar registros duplicados
df = df.drop_duplicates()


# Verificar el nuevo tamaño del dataset
print("Nuevo tamaño del dataset:", df.shape)


# Comprobar que ya no existan registros duplicados
duplicados_despues = df.duplicated().sum()

print("Registros duplicados después de limpiar:", duplicados_despues)
```

```python
# Crear característica: antiguedad
# Representa la cantidad de años completos que el colaborador
# lleva trabajando en la organización

fecha_referencia = pd.Timestamp("2026-08-09")

df["antiguedad"] = (
    fecha_referencia.year - df["fecha_ingreso"].dt.year
)

# Restar un año si todavía no ha cumplido su aniversario laboral
df["antiguedad"] = df["antiguedad"] - (
    (fecha_referencia.month < df["fecha_ingreso"].dt.month) |
    (
        (fecha_referencia.month == df["fecha_ingreso"].dt.month) &
        (fecha_referencia.day < df["fecha_ingreso"].dt.day)
    )
)

# Verificar resultados
df[["fecha_ingreso", "antiguedad"]].head()
```

```python
# Crear característica: capacitaciones_recibidas
# Se estima una capacitación por cada 8 horas de formación registradas

df["capacitaciones_recibidas"] = (
    df["horas_capacitacion"] / 8
).round().astype("Int64")

# Verificar resultados
df[
    [
        "horas_capacitacion",
        "capacitaciones_recibidas"
    ]
].head()
```

```python
# Crear característica: nivel_desempeno_calculado
# Clasifica el puntaje de desempeño en Bajo, Medio o Alto

def clasificar_desempeno(puntaje):
    if puntaje < 60:
        return "Bajo"
    elif puntaje < 80:
        return "Medio"
    else:
        return "Alto"


df["nivel_desempeno_calculado"] = (
    df["puntaje_desempeno"].apply(clasificar_desempeno)
)

# Verificar resultados
df[
    [
        "puntaje_desempeno",
        "nivel_desempeno_calculado"
    ]
].head()
```

```python
# Crear característica: riesgo_rotacion
# Se calcula utilizando satisfacción laboral, intención de permanecer
# y cantidad de días de ausencia

def calcular_riesgo_rotacion(fila):

    puntuacion_riesgo = 0

    # Baja intención de permanecer aumenta el riesgo
    if fila["respuestas_intencion_de_permanecer"] <= 2:
        puntuacion_riesgo += 2
    elif fila["respuestas_intencion_de_permanecer"] == 3:
        puntuacion_riesgo += 1

    # Baja satisfacción laboral aumenta el riesgo
    if fila["respuestas_satisfaccion_laboral"] <= 2:
        puntuacion_riesgo += 2
    elif fila["respuestas_satisfaccion_laboral"] == 3:
        puntuacion_riesgo += 1

    # Muchas ausencias pueden aumentar el riesgo
    if fila["ausencias_dias"] >= 8:
        puntuacion_riesgo += 2
    elif fila["ausencias_dias"] >= 4:
        puntuacion_riesgo += 1

    # Clasificar el riesgo obtenido
    if puntuacion_riesgo >= 4:
        return "Alto"
    elif puntuacion_riesgo >= 2:
        return "Medio"
    else:
        return "Bajo"


df["riesgo_rotacion"] = df.apply(
    calcular_riesgo_rotacion,
    axis=1
)

# Verificar resultados
df[
    [
        "respuestas_satisfaccion_laboral",
        "respuestas_intencion_de_permanecer",
        "ausencias_dias",
        "riesgo_rotacion"
    ]
].head()
```

```python
# Crear característica: estado_colaborador
# Si existe fecha de salida o el estado laboral indica salida,
# el colaborador se considera inactivo

def determinar_estado_colaborador(fila):

    # Si tiene fecha de salida registrada
    if pd.notna(fila["fecha_salida"]):
        return "Inactivo"

    # Si continúa laboralmente activo
    if str(fila["estado_laboral"]).strip().lower() == "activo":
        return "Activo"

    # Para cualquier otro estado laboral
    return "Inactivo"


df["estado_colaborador"] = df.apply(
    determinar_estado_colaborador,
    axis=1
)

# Verificar resultados
df[
    [
        "estado_laboral",
        "fecha_salida",
        "estado_colaborador"
    ]
].head()
```

```python
# Visualizar las nuevas características creadas

df[
    [
        "antiguedad",
        "capacitaciones_recibidas",
        "nivel_desempeno_calculado",
        "riesgo_rotacion",
        "estado_colaborador"
    ]
].head(10)
```
